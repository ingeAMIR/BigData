## Actividad 3 — Estadística descriptiva

**Propósito:** Aplicar medidas estadísticas básicas a una variable poblacional.

**Instrucciones**

Se selecciona un año del periodo histórico (2023, el más reciente disponible) y se filtran los datos correspondientes. Sobre la variable de **cambio anual de población** (`Annual change in population`) se calculan las medidas estadísticas descriptivas.

```python
anio = 2023
df_anio = df[df["Year"] == anio]
df_anio.head()
```

```python
variable = "Annual change in population"

media = df_anio[variable].mean()
mediana = df_anio[variable].median()
varianza = df_anio[variable].var()
desviacion_estandar = df_anio[variable].std()
minimo = df_anio[variable].min()
maximo = df_anio[variable].max()

print("Media:", media)
print("Mediana:", mediana)
print("Varianza:", varianza)
print("Desviación estándar:", desviacion_estandar)
print("Mínimo:", minimo)
print("Máximo:", maximo)
```

```python
entidad_min = df_anio.loc[df_anio[variable].idxmin(), "Entity"]
entidad_max = df_anio.loc[df_anio[variable].idxmax(), "Entity"]

print("Entidad con el valor mínimo:", entidad_min)
print("Entidad con el valor máximo:", entidad_max)
```

**Preguntas**

1. **¿Cuál es el valor de la media?**
   Para 2023, la media del cambio anual de población es de aproximadamente 2,008,098 personas. Este número está bastante inflado porque el dataset mezcla países individuales con agregados regionales (como continentes), que suman muchísima más gente que un país promedio.

2. **¿Cuál es el valor de la mediana?**
   La mediana es de 46,865.5 personas, mucho más baja que la media. Esa diferencia tan grande entre media y mediana es una señal clara de que la distribución está muy sesgada por esos valores extremos de las regiones agregadas.

3. **¿Cuál es el valor de la desviación estándar?**
   La desviación estándar es de aproximadamente 9,078,963 personas, un valor enorme que confirma la gran dispersión de los datos: hay desde países que pierden población hasta regiones enteras que ganan decenas de millones de habitantes en un solo año.

4. **¿Qué entidad presenta el valor mínimo y cuál el máximo?**
   Ucrania (Ukraine) presenta el valor mínimo, con una caída de población de más de 3.3 millones de personas en 2023, probablemente reflejo del conflicto y la migración forzada de esos años. El valor máximo corresponde a "Less developed regions, excluding China", una categoría agregada de la ONU que agrupa a muchos países en desarrollo, por lo que su crecimiento (más de 72 millones) es la suma de decenas de naciones y no el de un solo país.
